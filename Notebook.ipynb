# Trial durations
Three synthetic readings. The chart and output are saved in this notebook.

```python
from PIL import Image, ImageDraw
from IPython.display import display

samples = [("A", 24, "#087f73"), ("B", 43, "#cf5b2a"), ("C", 65, "#5460ad")]
print("Trial durations (ms):", [duration for _, duration, _ in samples])
chart = Image.new("RGB", (280, 116), "#f3f6f8")
draw = ImageDraw.Draw(chart)
draw.text((8, 4), "Trial duration (ms)", fill="#182d38")
for index, (trial, duration, color) in enumerate(samples):
    left = 28 + 82 * index
    draw.rectangle((left, 90 - duration, left + 40, 90), fill=color)
    draw.text((left, 94), f"{trial}: {duration}", fill="#182d38")
display(chart)
```
